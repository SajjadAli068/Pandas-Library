#### Merge in Pandas is used to combine two or more DataFrames based on a common column or key, similar to SQL joins.

```python
import pandas as pd 
df1 =pd.DataFrame({
    'city':['new york','chicago','rolando'],
    'temperature':[21,14,35]
})
df1
```

```python
df2 = pd.DataFrame({
    'city':['new york','chicago','rolando'],
    'humidity':[65,68,75]
})
df2
```

```python
df3 = pd.merge(df1,df2,on='city')
df3
```

```python
df1 =pd.DataFrame({
    'city':['new york','chicago','rolando' , 'baltimore'],
    'temperature':[21,14,35,32]
})
df1
```

```python
df2 = pd.DataFrame({
    'city':['new york','chicago','san francisco'],
    'humidity':[65,68,71]
})
df2
```

```python
df3=pd.merge(df1,df2,on='city')
df3
```

```python
# outer join
df3=pd.merge(df1,df2,on='city' , how='outer')
df3
```

```python
# how is inner join
df3 = pd.merge(df1,df2,on='city' , how='inner')
df3
```

```python
# Left join
df3 = pd.merge(df1,df2,on='city' , how='left')
df3
```

```python
# Right join
df3 = pd.merge(df1,df2,on='city' , how='right')
df3
```

```python
# indicater 
df3 = pd.merge(df1,df2,on='city' ,how='outer',indicator=True)
df3
```

```python
# suffixes 
df1 = pd.DataFrame({
    'city': ['new york','chicago','orlando','baltimore'],
    'tempurature':[21,14,35,38],
    'humidity':[65,68,71,75]
})
df1
```

```python
df2 = pd.DataFrame({
    'city':['chicago','new york','san diego'],
    'temperature':[21,14,35],
    'humidity':[65,68,71]
})
df2
```

```python
df3 = pd.merge(df1,df2,on='city' ,suffixes=('_left','_right'))
df3
```
