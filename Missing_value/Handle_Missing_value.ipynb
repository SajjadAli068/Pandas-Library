```python
import pandas as pd
```

```python
df = pd.read_csv('weather_data.csv',parse_dates=['day'])
df
```

```python
df.set_index('day' , inplace=True)
```

```python
df.isnull().sum()
```

```python
# fillna method very powerfull all empty cell fill within second 
df.fillna(0)
```

```python
# we use distionary to fill empty cell
df.fillna({
    'temperature':0,
    'windspeed ':0,
    'event': 'no event'
})
```

```python
# we use fillna metho ffill . first down fill empty cell
df.fillna(method='ffill')
```

```python
# we use fillna method bfill down to up method
df.fillna(method='bfill')
```
