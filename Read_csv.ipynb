```python
# if we remove header so we write Header = none and
# give them names use argument name
df=pd.read_csv('stock_market_data.csv',header=None , names=['col_a','col_b','col_c','col_d','col_e'] )
df
```

```python
# we are going orginal data and limit row we so we write nrows 
df = pd.read_csv('stock_market_data.csv',header=1,nrows=3)
df
```

#### Now we use na_value for empty cell

```python
df = pd.read_csv('stock_market_data.csv', header=1 , na_values=[-1,'not available'])
df
```

```python
# na_value use like distionary so we select column empty

df = pd.read_csv('stock_market_data.csv',header=1,na_values={
    'eps':['not available', -1],
    'revenue':[-1],
    'price':['n.a.'],
    'people':['n.a.']
})
df
```

```python
# to save csv and remove index
df.to_csv('no_index.csv',index=False)
```
